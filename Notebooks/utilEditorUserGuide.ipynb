## 🎯 Detailed Usage

### Opening the Editor

```python
# Basic usage
editor = AssetEditor('llcAssets_WBGroupLLC.json')
editor.start()

# Custom port and height
editor = AssetEditor('llcAssets_WBGroupLLC.json')
editor.start(port=5001, height=1000)
```

### Editing an Asset

1. Click the ✏️ (edit) button on any row
2. The edit dialog opens with all fields
3. Modify any field:
   - **Simple fields**: aID, propNm, dt, amt, aType, etc.
   - **JSON fields**: stakeholderPct, kwList
4. Click "Save" to update
5. Or click "Edit JSON" for raw JSON editing

### Adding a New Asset

1. Click "➕ Add Asset" button
2. Fill in the form (all fields are optional)
3. For JSON fields (stakeholderPct, kwList), use proper JSON syntax:
   ```json
   {"o20250801_1": 100.0}
   ```
4. Click "Save"

### Searching Assets

Type in the search box to filter by any field:
- Search by `aID`: "a20250826"
- Search by property: "High Mesa"
- Search by description: "Purchase"
- Search by amount: "219000"

### Saving to File

**Important**: Changes are only in memory until you click "💾 Save to File"

1. Make your edits (add, edit, delete)
2. Click "💾 Save to File" button in the header
3. Confirmation message appears
4. Your JSON file is now updated

### Raw JSON Editing

For advanced users:

1. Click ✏️ to edit an asset
2. Click "📝 Edit JSON" button
3. Edit the complete JSON structure
4. Click "💾 Save JSON"
5. Invalid JSON will show an error

## ⚠️ Important Notes

### Saving Changes
- **In-memory only** until you click "Save to File"
- The "Unsaved Changes" indicator shows if you have unsaved edits
- Always save before closing the notebook!

### Stopping the Server
- The Flask server runs in a background thread
- It will stop when you restart the Jupyter kernel
- Or when you close the notebook

### JSON Validation
- Simple fields (aID, propNm, etc.) are validated automatically
- JSON fields (stakeholderPct, kwList) must be valid JSON
- Invalid JSON will show an error and prevent saving

### Thread Safety
- Only one editor instance should run at a time
- If you need multiple editors, use different ports

```python
# Access the data directly
editor = AssetEditor('llcAssets_WBGroupLLC.json')
print(f"Loaded {len(editor.assets)} assets")

# Modify programmatically
editor.assets[0]['amt'] = 1000.0

# Save
editor.save_assets()
```

### 4. Quick Statistics
```python
editor = AssetEditor('llcAssets_WBGroupLLC.json')

total = sum(asset['amt'] for asset in editor.assets)
print(f"Total amount: ${total:,.2f}")

types = {}
for asset in editor.assets:
    atype = asset.get('aType', 'Unknown')
    types[atype] = types.get(atype, 0) + 1

print(f"Asset types: {types}")
```

## 🐛 Troubleshooting

### Port Already in Use
```
Error: Address already in use
```
**Solution**: Use a different port
```python
editor.start(port=5001)
```

### JSON Parse Error
```
Error: Invalid JSON in stakeholderPct
```
**Solution**: Check JSON syntax:
- Use double quotes: `{"key": "value"}`
- Not single quotes: `{'key': 'value'}`

### Module Not Found
```
ImportError: No module named 'flask'
```
**Solution**: Install Flask
```bash
pip install flask
```

### Changes Not Saving
**Solution**: Make sure to click "💾 Save to File" button

### IFrame Not Displaying
**Solution**: 
- Check if Jupyter trusts the notebook
- Restart Jupyter kernel
- Try a different browser
