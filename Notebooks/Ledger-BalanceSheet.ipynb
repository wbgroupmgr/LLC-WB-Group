```python
# llcBooks
import pandas as pd
from IPython.display import display, HTML

class llcBooks(object):
    '''
    Common core services of all books
    Each book contributed to the balance sheet
    Assets = Liabilities + Equity

    -> Depreciation: Shows non-cash expense reducing taxable income, lowering the building's book value (not market value).
    -> Snapshot in Time: Reflects finances on a specific date (e.g., Dec 31st).
    -> Investment Health: Reveals equity growth from paying down debt and property appreciation (on paper). 
    '''
    def __init__(self, llc, **kwargs):
        self.oID = self.__class__.__name__
        self.debug = kwargs.get('debug', False)
        self.objName = llc.objName
        self.llc = llc


    def display_dfs_side_by_side(self, dfs, titles=None):
        """
        Displays a list of pandas DataFrames horizontally using HTML flexbox.
        
        Args:
            dfs (list): A list of DataFrames to display.
            titles (list, optional): A list of titles for each DataFrame.
        """
        html_str = '<div style="display:flex;">'
        for i, df in enumerate(dfs):
            title_html = f'<h3>{titles[i]}</h3>' if titles else ''
            html_str += f'<div style="margin-right: 20px;">{title_html}{df.to_html()}</div>'
        html_str += '</div>'
        display(HTML(html_str))

    def displayBalSheeet(self, a, l, e):
        # Display ROW 1
        self.display_dfs_side_by_side([a.df, l.df], titles=['Assets', 'Liabilituy'])
        # Display ROW 2
        self.display_dfs_side_by_side([e.df], titles=['Equity'])
        
```

```python
# Assets
class llcAssets(llcBooks):
    '''
    --- What is owned  at a point in time ----
    -> Current Assets
       -> Cash / Petty Cash
       -> Real Estate / Inventory
       -> Acct Receivables
       -> Supply
       -> PrePaid Insurane
    -> Investments
    -> Property plant and equipment
       -> Land
       -> Rental Property
       -> Equipment
    -> Intangible assets
       -> Goodwill
       -> Trade names
    '''
    def __init__(self, llc, **kwargs):
        super().__init__(llc, **kwargs)

a = llcAssets(llc, debug=llc.debug)

    
```

```python
# Liabilities

class llcLiabilities(llcBooks):
    '''
    --- What is owed  at a point in time ---
    -> Mortgage/Notes Payable
    -> Accounts Payable
    -> Security/Pet Deposits (held for tenants)
    -> Accrued Expense - payable on obligations
    -> Wages payable
    -> Interest payable
    -> Tax payable
    -> Unearned revenue
    -> Unearned revenue
    '''
    def __init__(self, llc, **kwargs):
        super().__init__(llc, **kwargs)
l = llcLiabilities(llc, debug=llc.debug)
```

```python
# OwnerEquitiy

class OwnerEquity(llcBooks):
    '''
    --- owner's stake (Equity from investments/Profits) at a point in time ---
    -> Initial Investments / CommonStock :: % of investment
    -> Retained Earnings
    -> Accumulated depreciation
    -> Less: Treasury Stock
    '''
    def __init__(self, llc, **kwargs):
        super().__init__(llc, **kwargs)

e = OwnerEquity(llc)
a = e.llc.assets()
a
```
